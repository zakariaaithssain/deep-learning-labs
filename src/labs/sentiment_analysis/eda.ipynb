### exploratory data analysis

```python
import pandas as pd 
import random as rd

DATA_PATH = "data/imdb_reviews.csv"

data = pd.read_csv(DATA_PATH)

#random samples
text_samples = [data.iloc[i, 0] for i in rd.sample(range(data.shape[0]), 7)]

for sample in text_samples: 
    print(sample, end="\n\n")
```

from the text samples, we see that we need to remove:  
- HTML tags  
- numbers  
- punctuation  
- capitalization  
- replace extra whitespaces by only one  
- remove common words (the, is, ...)
