### Exercise 2: 

question 1:

```python
import unicodedata

def remove_accents(text):
    # NFD decomposition: 'è' → 'e' + '`' (base letter + combining accent)
    decomposed = unicodedata.normalize('NFD', text)
    # Filter out combining characters (category 'Mn' = Mark, nonspacing)
    result = ''.join(char for char in decomposed if unicodedata.category(char) != 'Mn')
    return result.lower()

```

question 2: 

```python
def remove_punct(text): 
    punct = "!?.,;:+-#$%&*()[]'"
    for sym in punct: 
        text = text.replace(sym, '')
    return text
```

for this corpus and task, coding numbers into their languistic format is the best approach as they represent durations and some basic metrics.

question 3: 

```python
import re

def remove_extra_spaces(text):
    return re.sub(r'\s+', ' ', text).strip()


print("5th sentence before: ", corpus[4])
print("5th sentence after: ", remove_extra_spaces(corpus[4]))
```

question 4: 

```python
simple_dict = {"tt": "traitement", "bcp": "beaucoup", "ras": "rien a signaler", "l’":"le"}
def expand_abbrev(text, dico): 
    words = text.split(" ")
    for word in words: 
        if word.lower() in dico: 
            words[words.index(word)] = dico[word]
    return ' '.join(words) 
```

question 5: 

```python
def normalize(text): 
    text = remove_accents(text)
    text = remove_punct(text)
    text = remove_extra_spaces(text)
    text = expand_abbrev(text, simple_dict)
    return text

normalized_corpus = [normalize(text) for text in corpus]
print("corpus before:\n ", corpus)
print("corpus after:\n ", normalized_corpus)
```

question 4:

```python
from os import setns


def pipeline_nlp(corpus):
    nlp = spacy.load('fr_core_news_sm')
    result = []
    for sentence in corpus: 
        doc = nlp(sentence)
        result.append([token.lemma_ for token in doc 
                    if not token.is_stop and not token.is_punct])
    return result

processed_corpus = pipeline_nlp(corpus)
for sentence in processed_corpus: 
    print(sentence)
```

## Synthese:

B1: 

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd 
import numpy as np


tfidf = TfidfVectorizer(input='content')
corpus = [' '.join(tokens) for tokens in processed_corpus]
tfidf_matrix = tfidf.fit_transform(corpus)

unique_words = np.array(tfidf.get_feature_names_out())
mean_tfidf = np.asarray(tfidf_matrix.mean(axis=0)).ravel()
top_5_corpus_indices = mean_tfidf.argsort()[-5:][::-1]

print("top 5 most discriminant terms across the entire dataset:")
for idx in top_5_corpus_indices:
    print(f"- {unique_words[idx]}: {mean_tfidf[idx]:.4f}")
```
