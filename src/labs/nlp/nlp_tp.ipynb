### Exercise 2: 

question 1:

```python
import unicodedata

def remove_accents(text):
    # NFD decomposition: 'è' → 'e' + '`' (base letter + combining accent)
    decomposed = unicodedata.normalize('NFD', text)
    # Filter out combining characters (category 'Mn' = Mark, nonspacing)
    result = ''.join(char for char in decomposed if unicodedata.category(char) != 'Mn')
    return result.lower()

```

question 2: 

```python
def remove_punct(text): 
    punct = "!?.,;:+-#$%&*()[]"
    for sym in punct: 
        text = text.replace(sym, '')
    return text
```

for this corpus and task, coding numbers into their languistic format is the best approach as they represent durations and some basic metrics.

```python
import re

def remove_extra_spaces(text):
    return re.sub(r'\s+', ' ', text).strip()

normalized_corpus = [remove_extra_spaces(text) for text in normalized_corpus]
print("5th sentence before: ", corpus[4])
print("5th sentence after: ", normalized_corpus[4])
```

question 4: 

```python
simple_dict = {"tt": "traitement", "bcp": "beaucoup", "ras": "rien a signaler"}
def expand_abbrev(text, dico): 
    words = text.split(" ")
    for word in words: 
        if word in words:
            if word in dico: 
                words[words.index(word)] = dico[word]
    return ' '.join(words) 
```

question 5: 

```python
def normalize(text): 
    text = remove_accents(text)
    text = remove_punct(text)
    text = remove_extra_spaces(text)
    text = expand_abbrev(text, simple_dict)
    return text

normalized_corpus = [normalize(text) for text in corpus]
print("corpus before:\n ", corpus)
print("corpus after:\n ", normalized_corpus)
```
